```python
import pandas as pd

url = "https://raw.githubusercontent.com/anisabendada/parent-health-assistant/main/data/test_set.csv"
tests = pd.read_csv(url)

print(tests.shape)
print(tests["type"].value_counts())
tests.head()
```

```python
trace_url = "https://raw.githubusercontent.com/anisabendada/parent-health-assistant/main/traceability.csv"
trace = pd.read_csv(trace_url)

covered = set()
for ids in trace["test_ids"]:
    for i in str(ids).split(","):
        covered.add(int(i))

missing = set(tests["id"]) - covered
print("Covered:", len(covered), "of", len(tests))
print("Not covered by any requirement:", missing)
```
